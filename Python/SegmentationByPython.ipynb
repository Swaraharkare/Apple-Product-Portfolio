```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv('AppleProductsCleaned.csv',header=None)
df.columns=['product_id','category','model_name','release_year'
      ,'price','rating','color','storage','ram','cpu' ,'gpu','battery_mAh','warranty','country','in_stock']
df
```

```python
df_sorted =df.sort_values(by='price')
df_sorted 
```

```python
# 2. Establish Percentile-Based Price Thresholds
# 33rd Percentile defines the ceiling for Budget
# 66th Percentile defines the floor for High
p33 = df_sorted['price'].quantile(0.33)
p66 = df_sorted['price'].quantile(0.66)
```

```python
def assign_price_tier(price):
    if price<=p33:
        return "Budget"
    elif price<=p66:
        return "Medium"
    else:
        return "High"

df['price_tier']=df_sorted['price'].apply(assign_price_tier)
df
```

```python
#Clean the RAM Column
df['ram'] = df['ram'].str.replace('GB', '', regex=False).astype(int)
df
```

```python
#Extract Processor Type
df['cpu_series']=np.where(df['cpu'].str.contains(r'\bM\d|M-',case=False,na=False),'M-series','A-series')
df
```

```python
#Bin RAM Capacities
boundaries=[0, 8, 24, np.inf]
labels=['Low (≤8GB)', 'Mid (12-24GB)', 'High (≥32GB)']
df['ram_cat']=pd.cut(df['ram'],bins=boundaries, labels=labels)
df
```

```python
# 5. Execute Multi-Dimensional Hardware Capability Pivot Table
hardware_matrix = pd.pivot_table(
    df,
    values=['product_id', 'price', 'rating', 'in_stock'],
    index=['cpu_series'],
    columns=['ram_cat'],
    aggfunc={
        'product_id': 'count',
        'price': 'median',
        'rating': 'mean',
        'in_stock': 'count'
    },
    fill_value=0
)
hardware_matrix
```
